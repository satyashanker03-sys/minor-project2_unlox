```python
#UNIVARIATE ANALYSIS
numeric_columns = ['latitude', 'longitude', 'price', 'minimum_nights', 'number_of_reviews', 'reviews_per_month', 'calculated_host_listings_count', 'availability_365']
for col in numeric_columns:
    plt.figure(figsize=(18, 10))
    plt.title("UNIVARIATE ANALYSIS")
    sns.histplot(df[col],kde=True,bins=30,color="blue")
    plt.show()
```

```python
#DROPPING THE LAST REVIEW COLUMN
df.drop(columns=['last_review'],inplace = True)
```

```python
#DROPPING THE NEIGHBOUR COLUMN
df.drop(columns=['neighbourhood'], inplace=True)
```

```python
#DROPPING THE NAME COLUMN
df.drop(columns=['name'],inplace=True)
```

```python
#DROPPING THE HOST NAME COLUMN
df.drop(columns=['host_name'],inplace =True)
```

```python
#DROPPING THE ID AND HOST ID COLUMN
cols_to_drop = ['id','host_id']
df = df.drop(columns=cols_to_drop, errors='ignore')
```

```python
#CHECKING FOR OUTLIERS IN THE NUMERICAL COLUMNS USING BOX PLOT
for col in numeric_columns:
    plt.figure(figsize=(18, 10))
    plt.title("CHECKING FOR OUTLIERS")
    sns.boxplot(x=df[col], color="blue")
    plt.show()
```

```python
#TREATING THE OUTLIER 
for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    df[col] = np.where(
        df[col] > upper_limit, upper_limit,
        np.where(df[col] < lower_limit, lower_limit, df[col])
    )
```

LINEAR REGRESSION :-

```python
#LINEAR REGRESSION
from sklearn.model_selection import train_test_split
```

```python
X = df.drop("price",axis = 1)
Y = df["price"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split( X, Y,test_size=0.20,random_state=69)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, Y_train)
print("model trained successfully")
```

```python
Y_pred = model.predict(X_test)
print("predict price : ")
print(Y_pred[:10])
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
#CALCULATING EVALUATION MATRICS
r2 = r2_score(Y_test, Y_pred)
mae = mean_absolute_error(Y_test, Y_pred)
mse = mean_squared_error(Y_test, Y_pred)
rmse = np.sqrt(mse)
```

```python
#DISPLAYING THE RESULTS
print("R-Squared Score:", r2)
print("Mean Absolute Error (MAE):", mae)
print("Mean Squared Error (MSE):", mse)
print("Root Mean Squared Error (RMSE):", rmse)
```

```python
#COMPAIRING TRAIN VS TEST SCORE
train_r2 = r2_score(Y_train, model.predict(X_train))
test_r2 = r2_score(Y_test, Y_pred)

print(f"Train R2 : {train_r2:.4f}")
print(f"Test R2  : {test_r2:.4f}")
```

```python
#PUSHING THE SCORE 
from sklearn.ensemble import RandomForestRegressor

rf = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
rf.fit(X_train, Y_train)

print(f"RF Train R2 : {r2_score(Y_train, rf.predict(X_train)):.4f}")
print(f"RF Test R2  : {r2_score(Y_test, rf.predict(X_test)):.4f}")
```

```python
#PLOTTING THE GRAPH BETWEEN ACTUAL AND PREDICTED PRICE 
plt.figure(figsize=(18, 10))
plt.scatter(Y_test, Y_pred, alpha=0.5)
plt.plot([min(Y_test), max(Y_test)],[min(Y_test), max(Y_test)],color='red',linestyle='--')
plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")
plt.title("Actual vs Predicted Airbnb Prices")
plt.show()
```

```python
#PLOTTING THE RESIDUAL GRAPH
residuals = Y_test - Y_pred
plt.figure(figsize=(18,10))
plt.scatter(Y_pred, residuals)
plt.axhline(y=0, color='red', linestyle='--')
plt.xlabel("Predicted Prices")
plt.ylabel("Residuals")
plt.title("Residual Plot")
plt.show()
     
```

```python
from sklearn.metrics import mean_absolute_error

results = pd.DataFrame({
    'Actual': Y_test,
    'Predicted': Y_pred
})

median_price = results['Actual'].median()

cheap = results[results['Actual'] <= median_price]
expensive = results[results['Actual'] > median_price]

print("MAE for Cheap Listings:",
      mean_absolute_error(cheap['Actual'], cheap['Predicted']))

print("MAE for Expensive Listings:",
      mean_absolute_error(expensive['Actual'], expensive['Predicted']))


     
```
